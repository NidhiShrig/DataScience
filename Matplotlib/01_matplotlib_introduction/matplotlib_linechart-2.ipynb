Line plot is most basic type of visualization in Matplotlib
It is used to display relationship between two different variable

The points are connected using straight lines

When Line Plots Are Used?
* Line plots are commonly used for:
* Time series data
* Trend analysis
* Stock prices
* Temperature changes
* Performance tracking

We can visualize this trend using line plot

Example:

Day	Temperature
1	30
2	32
3	31
4	35

```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([1,2,3,4])
y = np.array([30,32,31,35])

plt.plot(x,y)
plt.show()
```

```python
#lable is used describe what axis represents

plt.xlabel("days")
plt.ylabel("temperatures")
plt.show()
```

```python
#Title - A title describes what entire chart represents

plt.title("Temperature changes")
```

```python
#Grid- Add horizontal and vertical reference lines to chart
#This helps users:
# - read values more easily
# - compare points
# - understand trends

plt.grid()
```
