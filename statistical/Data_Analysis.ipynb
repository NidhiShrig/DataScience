Range = Difference between max and min value in dataset

#variance - Measures how far each data point is from the mean
v = Σ (X - M)**2/N
where,
X is each element in dataset
M is mean of the dataset
N is number of value in dataset

Larger variance indicates more spread out data points

#Standard devation: Square root of variance
- It represents the average distance of each data point from the mean
- Small SD -> valuse close to mean
- Large SD -> values widely spread

#percentile
40, 50, 60, 70, 80, 90, 100

If a student mark is at the 70th percentile
70 % student scored less than or equal to that mark

Percentage: Represents score out of 100
eg: 80% marks

percentile: Represents position in dataset
eg: 80th percentile ranking


How percentile is calculated

1. Organize the data into an ascending order
2. calculate the percentile location i by
      
       i = (p/100)*n

       p-> percentile of intrest
       i -> percentile location
       n -> no of data points

3. Determine the location by either a or b
 a. If i is a whole number - the pth percentile is the avg value at the ith location and value at i+1th location
 b. If i is not a whole number, the pth percentile is located at the whole no part of i+1

 eg: 10,15, 20, 30, 35, 60, 80

 25th percentile of data
 (25/100)*7 = 1.75 => the 25th percentile calculated from b is 15

Quartiles
Q1 ->  25th percentile
Q2 -> 50th percentile
Q3 -> 75th percentile

Outlier Formula(Values outside this range are considered outliers.)
Lower limit = Q1 − 1.5 × IQR
Upper limit = Q3 + 1.5 × IQR

Quartiles are used for:
✔ Detecting outliers
✔ Understanding data distribution
✔ Creating box plots
✔ Data skewness analysis


Visualization of box plot
|----|======|====|----|
Min  Q1    Q2   Q3   Max

Box = middle 50% of data
Line inside box = median




#IQR = Q3-Q1(This tells us how spread the middle portion of the data is)

```python
import numpy as np
data = [45,50,55,60,65,70,75,80,85]
q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 70)
print("Q1:", q1)
print("Q2:", q2)
print("Q3:", q3)
```

```python
import matplotlib.pyplot as plt

plt.boxplot(data)
plt.show()
```

#Box-Cox Transformation



```python
from scipy.stats import boxcox
```

```python
data = [2,3,4,5,6,8,10,12,50]
df = pd.DataFrame(data, columns=['usage'])
transformed_data, lam_value = boxcox(df['usage'])
print("Lambda:", lam_value)
df['boxcox_usage'] = transformed_data
```

skewness - Measures the asymmetry of data distribution
If data is perfectly balanced on both sides around mean no skweness
If data is tilted towards right -> positive skewness
If data is tilted towards left -> Negative skewness


No skewness:
Mean=Median=Mode

Positive skewness:
Mean>Median>Mode

Negative skewness
Mean< Median< Mode

Pearsons skweness coefficient:
3(Mean-Median)/standard deviation
Interpretation:
0-> symmetrical
>0 -> positive skewness
< 0 -> Negative skewness
